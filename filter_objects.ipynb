```python
from rdflib import Graph, URIRef
```

```python
INPUT_FILE = "mappingbased-objects_lang=en.ttl"
OUTPUT_FILE = "dbpedia_filtered.ttl"
```

```python
TARGET_RELATIONS = {
    "influenced",
    "influencedBy",
    "doctoralAdvisor",
    "doctoralStudent",
    "field",
}
```

```python
def get_relation_name(predicate):
    """
    Retorna apenas o nome da relação a partir da URI.

    Exemplo:
        http://dbpedia.org/ontology/influenced
    ->  influenced
    """
    return str(predicate).rsplit("/", 1)[-1]
```

```python
def get_relation_name(predicate):
    """
    Retorna apenas o nome da relação a partir da URI.

    Exemplo:
        http://dbpedia.org/ontology/influenced
    ->  influenced
    """
    return str(predicate).rsplit("/", 1)[-1]
```

```python
def filter_dbpedia():
    graph = Graph()

    print(f"Lendo: {INPUT_FILE}")

    # Parse do arquivo Turtle
    graph.parse(INPUT_FILE, format="turtle")

    print(f"Total de triplas: {len(graph):,}")

    # Novo grafo contendo apenas as relações desejadas
    filtered_graph = Graph()

    for subject, predicate, obj in graph:
        relation = get_relation_name(predicate)

        if relation in TARGET_RELATIONS:
            filtered_graph.add((subject, predicate, obj))

    print(f"Triplas filtradas: {len(filtered_graph):,}")

    # Salva o resultado
    filtered_graph.serialize(
        destination=OUTPUT_FILE,
        format="turtle"
    )

    print(f"Arquivo salvo em: {OUTPUT_FILE}")
```

```python
def filter_dbpedia(input_file, output_file):
    """
    Filtra um arquivo Turtle do DBpedia mantendo apenas
    as relações especificadas em TARGET_RELATIONS.

    O arquivo é processado linha a linha para evitar
    carregar todo o dataset na memória.
    """

    total_lines = 0
    selected_lines = 0

    with open(input_file, "r", encoding="utf-8") as fin, \
         open(output_file, "w", encoding="utf-8") as fout:

        for line in fin:
            total_lines += 1

            # Mantém linhas de prefixo/comentários
            if line.startswith("@prefix") or line.startswith("@base"):
                fout.write(line)
                continue

            # Verifica se alguma das relações aparece como predicado
            # na linha.
            if any(
                f"/{relation} " in line or
                f"/{relation}>" in line
                for relation in TARGET_RELATIONS
            ):
                fout.write(line)
                selected_lines += 1

            # Mostra progresso periodicamente
            if total_lines % 1_000_000 == 0:
                print(
                    f"Linhas processadas: {total_lines:,} | "
                    f"Linhas selecionadas: {selected_lines:,}"
                )

    print("\nFinalizado!")
    print(f"Linhas processadas: {total_lines:,}")
    print(f"Linhas selecionadas: {selected_lines:,}")
    print(f"Arquivo gerado: {output_file}")
```

```python
filter_dbpedia(INPUT_FILE, OUTPUT_FILE)
```
